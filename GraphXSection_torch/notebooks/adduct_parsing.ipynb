```python
import sys
from pathlib import Path
import pandas as pd

# Add the parent of project root to the path so GraphXSection_torch can be imported as a module
project_root = Path.cwd().parent
parent_dir = project_root.parent
if str(parent_dir) not in sys.path:
    sys.path.insert(0, str(parent_dir))

import GraphXSection_torch

from GraphXSection_torch.adduct_features import AdductFeatureExtractor
```

```python
data_sigmaccs = pd.read_excel(
    "/mnt/c/Users/robbe/Work/Waters_GNN_SmalMol/data/SigmaCCS/42004_2023_939_MOESM3_ESM.xlsx",
    header=1,
)
ccs_consolidated = pd.read_csv(
    "/mnt/c/Users/robbe/Work/Waters_GNN_SmalMol/data/CCS_consolidated_SMILES_processed.csv"
)

data_graphccs = pd.read_csv(
    "/mnt/c/Users/robbe/Work/Waters_GNN_SmalMol/data/GraphCCS.csv"
)
```

```python
adducts_graphccs = data_graphccs["Adduct"].unique().tolist()
adducts_consolidated = ccs_consolidated["Adduct"].unique().tolist()
```

```python
adducts_graphccs
```

```python
afe = AdductFeatureExtractor()

# Define feature names based on the feature composition
feature_names = [
    "charge_magnitude",
    "charge_positive",
    "charge_negative",
    "ion_H",
    "ion_Na",
    "ion_K",
    "ion_NH4",
    "ion_Li",
    "ion_other_metal",
    "ion_formate",
    "ion_halide",
    "ion_other_unknown",
    "mass_delta",
]

print("Adduct parsing and feature extraction for GraphCCS dataset:\n")
for adduct in adducts_graphccs:
    parsed_adduct = afe.parse_adduct(adduct)
    features = afe.extract_features(adduct)
    print(f"Adduct: {adduct}")
    print(f"Parsed: {parsed_adduct}")
    print("Features:")
    for name, value in zip(feature_names, features):
        print(f"  {name}: {value}")
    print("\n" * 2)

print("===============================\n")
print("Adduct parsing and feature extraction for consolidated CCS dataset:\n")
for adduct in adducts_consolidated:
    parsed_adduct = afe.parse_adduct(adduct)
    features = afe.extract_features(adduct)
    print(f"Adduct: {adduct}")
    print(f"Parsed: {parsed_adduct}")
    print("Features:")
    for name, value in zip(feature_names, features):
        print(f"  {name}: {value}")
    print("\n" * 2)
```
